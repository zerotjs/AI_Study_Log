# 연산자
- 산술 연산자(숫자 연산)
    - +(덧셈),-(뺄셈), *(곱셈), **(거듭제곱), /(나눗셈), //(몫), %(나머지)
- 대입 연산자(대입 및 복합 대입)
    - =(변수 저장), +=(변수에 우변을 덧셈한 값 저장), -=(변수에 우변을 뺄셈한 값 저장), *=(변수에 우변을 곱한 값 저장), **=(변수에 우변을 거듭제곱한 값 저장), /=(변수에 우변을 나눈 값 저장), //=(변수에 우변을 나눈 몫 저장), %=(변수에 우변을 나눈 나머지 저장)
- 관계 연산자(크기 비교)
    - 2개의 항을 비교하여 그 결과를 논리 자료형으로 반환
    - >, >=, <, <=, ==, !=
- 논리 연산자
    - and(모두 참이여야 True, 아니면 False), or(하나라도 참이면 True, 아니면 False), not
- 비트 연산자(이진 연산)
    - &(비트 AND 처리), |(비트 OR 처리), ^(비트 XOR 처리), ~(비트 NOT 처리), <<(비트로 변환하여 왼쪽으로 n비트 이동), >>(비트로 변환하여 오른쪽으로 n비트 이동)
    - 왼쪽 SHIFT : 비트 단위로 왼쪽으로 N비트 이동하면 2의 N 거듭제곱만큼 곱셈
    - 오른쪽 SHIFT : 비트 단위로 오른쪽으로 N비트 이동하면 2의 N 거듭제곱만큼 나눗셈
- 기타 연산자
    - 조건 연산
        - 참 if 조건식 else 거짓
    - 시퀀스 연산
        - +(연결), *(반복)
    - 멤버십 연산자
        - in(포함되어있으면 True), not in(포함되어 있지 않으면 True)

```python
# 기본 예제 연습

# 산술 연산자
print("산술 연산자")
a = 3
b = 5
print(f'{a}+{b}={a+b}')
print(f'{a}-{b}={a-b}')
print(f'{a}*{b}={a*b}')
print(f'{a}**{b}={a**b}')
print(f'{a}/{b}={a/b}')
print(f'{a}//{b}={a//b}')
print(f'{a}%{b}={a%b}')

# 대입 연산자
print("\n대입 연산자")
piggy_bank = 0

money = 10000
piggy_bank += money
print(f'저금통에 {money}원을 넣었습니다')
print(f'저금통에 {piggy_bank}원이 있습니다')

piggy_bank -= 1500
print('저금통에서 1500원을 꺼냈습니다')
print(f'저금통에 {piggy_bank}원이 있습니다')

# 관계 연산자
print("\n관계 연산자")
a = 15
print(f'{a}>10 : {a>10}')
print(f'{a}<10 : {a<10}')
print(f'{a}>=10 : {a>=10}')
print(f'{a}<=10 : {a<=10}')
print(f'{a}==10 : {a==10}')
print(f'{a}!=10 : {a!=10}')

# 논리 연산자
print("\n논리 연산자")

a = 10
b = 0
print(f'{a} > 0 and {b}>0 : {a>0 and b>0}')
print(f'{a} > 0 or {b}>0 : {a>0 or b>0}')
print(f'not {a} : {not a}')
print(f'not {b} : {not b}')
```

```python
# 비트 연산자
print("\n비트 연산자")
a = 10
b = 70
print(f'a & b : {a&b}')
print(f'a | b : {a|b}')
print(f'a ^ b : {a^b}')
print(f'~a : {~a}')
print(f'a << 1 : {a<<1}')
print(f'a >> 1 : {a>>1}')

# 시퀀스 연산자
print("\n시퀀스 연산자")
star = '*'
space = ' '
print(space * 4 + star * 1)
print(space * 3 + star * 3)
print(space * 2 + star * 5)
print(space * 1 + star * 7)
print(space * 0 + star * 9)

# 조건 연산자
print("\n조건 연산자")
a = 10
b = 20
result = (a-b) if (a>=b) else (b-a)
print(f'{a}와 {b}의 차는 {result}입니다')
```

```python
# 응용 예제
# 1번
print("1번")
a = int(input("10~99 사이의 정수를 입력하세요"))
print(f'십의 자리 : {a//10}')
print(f'일의 자리 : {a%10}')

# 2번
print("\n2번")
a = int(input("초를 입력하세요"))
b = a//3600
c = (a - b*3600) // 60
d = a - b*3600 - c*60
print(f"변환 결과는 {b}시간 {c}분 {d}초입니다.")

# 3번
print("\n3번")
a = int(input("4자리 사원번호를 입력하세요"))
result = "오전" if (a%10>=5) else "오후"
print(f'근무 시간은 {result}입니다')
```

```python
#4번
print("4번")
print("한 박스에 20개의 라면을 담을 수 있습니다.\n라면의 개수를 입력하시면 필요한 박스 수를 알려드립니다")
a = int(input("라면의 개수를 입력하세요"))
b = (a//20) if (a%20 ==0) else (a//20 + 1)
print(f'라면 {a}개를 담으려면 {b}박스가 필요합니다.')

#5번
print("\n5번")
a = int(input("국어 점수를 입력하세요"))
b = int(input("영어 점수를 입력하세요"))
c = int(input("수학 점수를 입력하세요"))
d = (a+b+c) /3
result = "합격" if (d>=80) else "불합격"
print(f'평균은 {d}점이고, 결과는 {result}입니다')
```
